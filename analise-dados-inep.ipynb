```python
import os, json, unicodedata
from collections import Counter
import pandas as pd

pd.options.mode.copy_on_write = True

# Ajuste para a raiz onde estão as pastas "microdados_censo_da_educacao_superior_{year}"
ROOT_DIR = r"C:\Users\joaod\OneDrive\Documentos\PROGESTAO\alvaro"
CACHE_DIR = os.path.join(ROOT_DIR, "_cache"); os.makedirs(CACHE_DIR, exist_ok=True)

# Quais anos você quer inspecionar
ANOS = list(range(2010, 2024))  # 2010..2023

# ------------ nomes possíveis de colunas (edite/complete à vontade) ------------
COLMAP_DEFAULT = {
  "CURSOS": {
    "ano":             ["NU_ANO_CENSO", "NU_ANO"],
    "co_ies":          ["CO_IES"],
    "sg_uf":           ["SG_UF"],
    "no_curso":        ["NO_CURSO", "NM_CURSO"],
    "tp_grau":         ["TP_GRAU_ACADEMICO", "DS_GRAU_ACADEMICO"],
    "tp_modalidade":   ["TP_MODALIDADE_ENSINO", "DS_MODALIDADE_ENSINO"],
    "qt_ing":          ["QT_ING", "QT_INGRESSO"],
    "qt_mat":          ["QT_MAT", "QT_MATRICULA"],
    "qt_conc":         ["QT_CONC", "QT_CONCLUINTE"],
    # áreas (opcionais)
    "area_geral":      ["NO_CINE_AREA_GERAL", "NO_GRANDE_AREA", "NO_CINE_AREA",
                        "NO_AREA_CURSO", "NO_CINE_AREA_DG", "NO_CINE_AREA_GERAL_CURSO"],
    "area_especifica": ["NO_CINE_AREA_ESPECIFICA", "NO_CINE_AREA_DETALHADA",
                        "NO_CINE_AREA_ESPECIFICA_CURSO", "NO_CINE_AREA_DETALHADA_CURSO"],
  },
  "IES": {
    "co_ies":   ["CO_IES"],
    "cat_adm":  ["TP_CATEGORIA_ADMINISTRATIVA", "DS_CATEGORIA_ADMINISTRATIVA"]
  }
}

# ------------ normalização e resolução de nomes ------------
def _norm(s: str) -> str:
    if s is None: return ""
    s = str(s).strip().upper()
    s = unicodedata.normalize("NFKD", s)
    return "".join(ch for ch in s if not unicodedata.combining(ch))

def resolve_colnames(available_cols, synonyms_list):
    """tenta: (1) exato; (2) normalizado; (3) contains normalizado."""
    if not synonyms_list: return None
    av_set = set(available_cols)

    # exato
    for c in synonyms_list:
        if c in av_set: return c

    # normalizado
    norm_map = {_norm(c): c for c in available_cols}
    for c in synonyms_list:
        nc = _norm(c)
        if nc in norm_map: return norm_map[nc]

    # contains normalizado (fallback)
    for c in synonyms_list:
        nc = _norm(c)
        hits = [orig for orig in available_cols if nc in _norm(orig)]
        if hits: return hits[0]
    return None

def build_path(year: int, kind: str) -> str:
    base = os.path.join(
        ROOT_DIR,
        f"microdados_censo_da_educacao_superior_{year}",
        f"microdados_censo_da_educacao_superior_{year}",
        "dados",
    )
    if kind == "CURSOS":
        return os.path.join(base, f"MICRODADOS_CADASTRO_CURSOS_{year}.CSV")
    elif kind == "IES":
        return os.path.join(base, f"MICRODADOS_ED_SUP_IES_{year}.CSV")
    else:
        raise ValueError("kind deve ser 'CURSOS' ou 'IES'.")

# ------------ mapeamentos de código -> rótulo (para ver o "mapeado único") ------------
def map_modalidade(x):
    if pd.isna(x): return None
    try:
        return {1: "Presencial", 2: "EAD"}[int(x)]
    except Exception:
        xs = _norm(x)
        if "PRESEN" in xs: return "Presencial"
        if "DIST" in xs or "A DIST" in xs or "EAD" in xs: return "EAD"
        return None

def map_grau(x):
    if pd.isna(x): return None
    try:
        return {1: "BACHARELADO", 2: "LICENCIATURA", 3: "TECNOLÓGICO"}[int(x)]
    except Exception:
        return _norm(x)

def map_cat_adm(x):
    if pd.isna(x): return None
    try:
        return {
            1: "Pública Federal", 2: "Pública Estadual", 3: "Pública Municipal",
            4: "Privada com fins", 5: "Privada sem fins"
        }[int(x)]
    except Exception:
        xs = _norm(x)
        if "FEDER" in xs: return "Pública Federal"
        if "ESTADU" in xs: return "Pública Estadual"
        if "MUNIC" in xs:  return "Pública Municipal"
        if "SEM FINS" in xs: return "Privada sem fins"
        if "COM FINS" in xs or "PRIVADA" in xs: return "Privada com fins"
        return None
```

```python
# %%  LISTAR VALORES ÚNICOS — EXECUÇÃO

def list_uniques_for_year(year: int,
                          colmap=COLMAP_DEFAULT,
                          chunksize=200_000,
                          preview=30):
    """
    Retorna dicionário com únicos de:
      CURSOS: [area_geral, area_especifica, no_curso, tp_grau(raw/mapped), tp_modalidade(raw/mapped), sg_uf]
      IES:    [cat_adm(raw/mapped)]
    Salva JSON completo em _cache/uniques_{year}.json
    """
    out = {"year": year, "CURSOS": {}, "IES": {}}

    # ---------- CURSOS ----------
    p_cursos = build_path(year, "CURSOS")
    head_c = pd.read_csv(p_cursos, sep=";", encoding="ISO-8859-1", nrows=0)
    cols_c = list(head_c.columns)

    def res(key): return resolve_colnames(cols_c, colmap["CURSOS"].get(key, []))

    c_area_g  = res("area_geral")
    c_area_e  = res("area_especifica")
    c_nome    = res("no_curso")
    c_grau    = res("tp_grau")
    c_mod     = res("tp_modalidade")
    c_uf      = res("sg_uf")

    usecols_c = [c for c in [c_area_g, c_area_e, c_nome, c_grau, c_mod, c_uf] if c]
    uniques_c = {c: set() for c in usecols_c}
    mapped_c  = {}

    if c_grau: mapped_c[c_grau] = set()
    if c_mod:  mapped_c[c_mod]  = set()

    for chunk in pd.read_csv(p_cursos, sep=";", encoding="ISO-8859-1",
                             usecols=usecols_c, chunksize=chunksize, low_memory=True):
        for c in usecols_c:
            vals = pd.unique(chunk[c].dropna())
            uniques_c[c].update(vals)

        if c_grau:
            mapped_c[c_grau].update(
                v for v in map(map_grau, pd.unique(chunk[c_grau].dropna())) if v is not None
            )
        if c_mod:
            mapped_c[c_mod].update(
                v for v in map(map_modalidade, pd.unique(chunk[c_mod].dropna())) if v is not None
            )

    # organiza saída CURSOS
    for c, s in uniques_c.items():
        vals_sorted = sorted([str(v) for v in s])
        out["CURSOS"][c] = {
            "unique_count": len(s),
            "sample_preview": vals_sorted[:preview],
            "all_values_sorted": vals_sorted,  # completo no JSON
        }
    for c, s in mapped_c.items():
        vals_sorted = sorted([str(v) for v in s])
        out["CURSOS"][f"{c}__MAPPED"] = {
            "unique_count": len(s),
            "sample_preview": vals_sorted[:preview],
            "all_values_sorted": vals_sorted,
        }

    # ---------- IES ----------
    p_ies = build_path(year, "IES")
    head_i = pd.read_csv(p_ies, sep=";", encoding="ISO-8859-1", nrows=0)
    cols_i = list(head_i.columns)

    c_cat = resolve_colnames(cols_i, colmap["IES"]["cat_adm"])
    usecols_i = [c_cat] if c_cat else []

    if usecols_i:
        uniq_i_raw = set()
        uniq_i_map = set()
        for chunk in pd.read_csv(p_ies, sep=";", encoding="ISO-8859-1",
                                 usecols=usecols_i, chunksize=100_000, low_memory=True):
            uniq_i_raw.update(pd.unique(chunk[c_cat].dropna()))
            uniq_i_map.update(
                v for v in map(map_cat_adm, pd.unique(chunk[c_cat].dropna())) if v is not None
            )

        vals_sorted = sorted([str(v) for v in uniq_i_raw])
        out["IES"][c_cat] = {
            "unique_count": len(uniq_i_raw),
            "sample_preview": vals_sorted[:preview],
            "all_values_sorted": vals_sorted,
        }
        vals_sorted_m = sorted([str(v) for v in uniq_i_map])
        out["IES"][f"{c_cat}__MAPPED"] = {
            "unique_count": len(uniq_i_map),
            "sample_preview": vals_sorted_m[:preview],
            "all_values_sorted": vals_sorted_m,
        }

    # ---------- salvar JSON ----------
    jpath = os.path.join(CACHE_DIR, f"uniques_{year}.json")
    with open(jpath, "w", encoding="utf-8") as f:
        json.dump(out, f, ensure_ascii=False, indent=2)

    # ---------- print resumo ----------
    print(f"\n=== {year} — CURSOS ===")
    for k, v in out["CURSOS"].items():
        print(f"{k}: {v['unique_count']} únicos | exemplo: {v['sample_preview'][:5]}")
    print(f"\n=== {year} — IES ===")
    for k, v in out["IES"].items():
        print(f"{k}: {v['unique_count']} únicos | exemplo: {v['sample_preview'][:5]}")
    print(f"\nArquivo salvo: {jpath}")
    return out

def list_uniques_for_years(years=ANOS, **kwargs):
    reports = {}
    for y in years:
        try:
            reports[y] = list_uniques_for_year(y, **kwargs)
        except Exception as e:
            print(f"[WARN] {y}: {e}")
    # salva índice de anos
    idx_path = os.path.join(CACHE_DIR, "uniques_index_years.json")
    with open(idx_path, "w", encoding="utf-8") as f:
        json.dump(sorted(list(reports.keys())), f, ensure_ascii=False)
    print(f"\nIndex salvo: {idx_path}")
    return reports

# ---------- EXECUTAR ----------
# reports = list_uniques_for_years([2021, 2022, 2023])  # exemplo
# reports[2023]["CURSOS"].keys()  # veja quais colunas foram encontradas no 2023
```

```python
import zipfile, glob, re
import pandas as pd

dfs = []
for z in glob.glob("dados/microdados_censo_da_educacao_superior_*.zip"):
    # pega ano do nome do zip
    m = re.search(r"(\d{4})", z)
    if not m:
        continue
    year = int(m.group(1))
    if year < 2012:
        continue  # pula anos anteriores

    with zipfile.ZipFile(z) as zf:
        # só arquivos CSV dentro da pasta dados/
        files = [f for f in zf.namelist() if "/dados/" in f and f.upper().endswith(".CSV")]
        files_sorted = sorted(files, key=lambda x: zf.getinfo(x).file_size, reverse=True)

        if len(files_sorted) < 2:
            print("⚠️ Menos de 2 CSVs em", z)
            continue

        target = files_sorted[1]  # 2º maior CSV
        with zf.open(target) as f:
            df_tmp = pd.read_csv(f, sep=";", low_memory=False, encoding="latin1")
            df_tmp["ANO_CENSO"] = year
            df_tmp["ARQUIVO_ORIGEM"] = target
            dfs.append(df_tmp)

df_ies = pd.concat(dfs, ignore_index=True)
df_ies
```
